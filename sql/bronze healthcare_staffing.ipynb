duplicates if any 

```sql
-- COMMAND ----------

SELECT
  'bronze_nurse_staffing' AS table_name,
  COUNT(*) AS row_count
FROM bronze_nurse_staffing

UNION ALL

SELECT
  'bronze_monthly_labor_cost',
  COUNT(*)
FROM bronze_monthly_labor_cost

UNION ALL

SELECT
  'bronze_monthly_patient_volume',
  COUNT(*)
FROM bronze_monthly_patient_volume;
```

check dates 

```sql
-- COMMAND ----------

SELECT
  'bronze_nurse_staffing' AS table_name,
  MIN(month) AS first_month,
  MAX(month) AS last_month,
  COUNT(DISTINCT month) AS month_count
FROM bronze_nurse_staffing

UNION ALL

SELECT
  'bronze_monthly_labor_cost',
  MIN(month),
  MAX(month),
  COUNT(DISTINCT month)
FROM bronze_monthly_labor_cost

UNION ALL

SELECT
  'bronze_monthly_patient_volume',
  MIN(month),
  MAX(month),
  COUNT(DISTINCT month)
FROM bronze_monthly_patient_volume;
```
